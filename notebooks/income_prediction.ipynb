```python
numeric_features=x_train.select_dtypes(include="number").columns
catagorical_features=x_train.select_dtypes(include="object").columns
```

```python
numeric_features
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
categorical_transformer= OneHotEncoder(handle_unknown="ignore")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
```

```python
preprocessor= ColumnTransformer(transformers=[
    ("num", StandardScaler(), numeric_features), 
    ("cat", categorical_transformer, catagorical_features)])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
baseline_model= Pipeline(steps=[
    ("prerocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))])
```

```python
baseline_model.fit(x_train, y_train)
```

```python
y_pred_BL= baseline_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_BL= accuracy_score(y_test, y_pred_BL)
```

```python
accuracy_BL
```

```python
from sklearn.metrics import classification_report
```

```python
BL_report= classification_report(y_test, y_pred_BL, output_dict=True)
```

```python
pd.DataFrame(BL_report)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test, y_pred_BL)
```

- Used OneHotEncoder and StandardScaler to prepare the data for the model
- Used ColumnTransformer to apply different transformations to different columns
- Used Pipeline to organize the preprocessing steps and the model
- Trained a LogisticRegression model as baseline model
- The model achieved 85% accuracy but since the target classes are imbalanced accuracy alone is not enough to evaluate the model
- The model performed better on the <=50K class because it had more training samples but its performance on the >50K class was lower

```python
balanced_weight_modeal= Pipeline(steps=[
    ("processor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000, class_weight="balanced"))])
```

```python
balanced_weight_modeal.fit(x_train, y_train)
```

```python
y_pred_BW= balanced_weight_modeal.predict(x_test)
```

```python
accuracy_BW= accuracy_score(y_test, y_pred_BW)
```

```python
accuracy_BW
```

```python
BW_report= classification_report(y_test, y_pred_BW, output_dict=True)
```

```python
pd.DataFrame(BW_report)
```

```python
confusion_matrix(y_test, y_pred_BW)
```

- Since the target classes were imbalanced I trained a Logistic Regression model with class_weight='balanced'
- In new model the recall for >50K improved, but precision decreased because the model gave more weight to minority class and that causes more false positive predictions
- Despite the decrease in precision the F1-score for >50K improved overall

```python
outlier_result= []
for col in numeric_features:
    q1= x_train[col].quantile(0.25)
    q3= x_train[col].quantile(0.75)
    iqr= q3-q1
    
    lower_bound= q1-1.5*iqr
    upper_bound= q3+1.5*iqr
    
    outliers= ((x_train[col] < lower_bound) | (x_train[col] > upper_bound))
    outlier_result.append({"Feature": col, "Q1": q1, "Q3": q3, "IQR": iqr, 
                            "Lower bound": lower_bound, "Upper bound": upper_bound,
                            "Outlier count": outliers.sum(),
                            "Outlier percentage": round(outliers.mean()*100, 2)})

pd.DataFrame(outlier_result)
```

Outlier detection and analysis:
- Visualized numerical features using histograms and boxplots
- Used IQR (Interquartile Range) method to identifyoutliers
- Analyzed the number and percentage of outliers in each numerical feature
- capital-loss and capital-gain have highly skewed distributions
- Decided not to remove outliers at this stage(extreme values are not necessarily invalid)
